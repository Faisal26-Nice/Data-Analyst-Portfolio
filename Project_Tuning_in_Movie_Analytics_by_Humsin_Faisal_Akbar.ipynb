```python
df_ratings["rating_time"] = pd.to_datetime(df_ratings["timestamp"],
                                           unit = "s")
```

```python
#Delete columns
df_ratings = df_ratings.drop(columns = ["timestamp"])
```

```python
df_metadata["release_date"] = pd.to_datetime(df_metadata["release_date"])
```

```python
df_metadata["release_year"] = df_metadata["release_date"].dt.year
```

```python
df_metadata.info()
```

```python
df_metadata["release_year"].describe()
```

```python
usecols = ["id", "imdb_id", "budget", "original_language", "runtime",
           "revenue", "genres", "release_year", "vote_count", "status"]
```

```python
movies_data = df_metadata[usecols]
movies_data.head()
```

```python
movies_data["status"].value_counts()
```

```python
movies_data[movies_data["release_year"].isna()]
```

```python
released_movies = movies_data[movies_data["status"] == 'Released']
released_movies = released_movies.drop(columns = "status")
```

```python
#Profit = revenue - budget
released_movies["profit"] = released_movies["revenue"] - released_movies["budget"]
```

```python
released_movies[released_movies["profit"] < 0].sort_values("profit")
```

```python
#Detecting missing values

released_movies[released_movies["release_year"].isna()]
```

```python
# Drop missing values
released_movies = released_movies.dropna(subset = ["release_year"])
released_movies["release_year"] = released_movies["release_year"].astype("int")
```

```python
#Check duplicate values
released_movies.duplicated().sum()
```

```python
released_movies[released_movies.duplicated()]
```

```python
released_movies = released_movies.drop_duplicates()
```

```python
movie_ratings = df_ratings.groupby("movieId", as_index = False)["rating"].agg(["mean", "count"])
```

```python
movie_ratings = movie_ratings.rename(columns = {
    "mean": "average_ratings",
    "count" : "user_count"
})
```

```python
released_movies_ratings = released_movies.merge(movie_ratings,
                      left_on = "id",
                      right_on = "movieId",
                      how = "inner")
```
